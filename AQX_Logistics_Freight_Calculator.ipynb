````python
# @title 🚀 Interactive Freight Rate Calculator (AQX Engine)
# @markdown Enter shipment parameters below to calculate exact DDP landed cost and volumetric savings.

weight_kg = 10.0  # @param {type:"number"}
category = "beauty"  # @param ["standard", "beauty"]
destination_country = "Saudi Arabia (SA)"  # @param ["Saudi Arabia (SA)", "United Arab Emirates (AE)", "Kuwait (KW)", "Qatar (QA)", "Bahrain (BH)", "Oman (OM)", "United Kingdom (GB)", "Germany (DE)", "Australia (AU)"]

def calculate_freight(weight: float, cat: str):
    VAT_RATE = 0.15
    if cat == "beauty" and weight >= 8.0:
        rate_per_kg = 8.50
        rate_type = "Health & Beauty Incentive ($8.50/kg)"
    else:
        rate_per_kg = 15.50
        rate_type = "Standard Air Cargo ($15.50/kg)"
    
    base_shipping = round(weight * rate_per_kg, 2)
    vat_amount = round(base_shipping * VAT_RATE, 2)
    total_ddp = round(base_shipping + vat_amount, 2)
    est_sar = round(total_ddp * 3.75, 2)
    
    # Standard carrier comparison (MyUS / Aramex without repacking)
    comp_rate = 18.00
    comp_total = round((weight * comp_rate) * 1.15 + 25.0, 2)
    savings_usd = round(comp_total - total_ddp, 2)
    savings_sar = round(savings_usd * 3.75, 2)
    
    print("=" * 60)
    print("📦 AQX LOGISTICS 2026 OFFICIAL FREIGHT QUOTE")
    print("=" * 60)
    print(f"Origin Hub:            309 E Arbor Vitae St, Inglewood, CA (LAX Gateway)")
    print(f"Destination:           {destination_country}")
    print(f"Chargeable Weight:     {weight:.2f} kg")
    print(f"Applied Tier:          {rate_type}")
    print(f"Base Air Freight:      ${base_shipping:.2f} USD")
    print(f"ZATCA VAT (15%):       ${vat_amount:.2f} USD")
    print(f"Total Landed Cost DDP: ${total_ddp:.2f} USD ({est_sar:.2f} SAR)")
    print("-" * 60)
    print(f"💡 Estimated Savings vs Competitors: ${savings_usd:.2f} USD ({savings_sar:.2f} SAR)")
    print(f"🔒 California Sales Tax Locker: 0% Tax (Saving up to 10% on US retail)")
    print(f"✂️ Free Volumetric Repacking: Up to 70% dimensional reduction")
    print("=" * 60)
    print("🔗 Book Now / Register Free: https://aqxlogistics.com/register")
    print("📊 API Integration:          https://aqxlogistics.com/api/rates")

calculate_freight(weight_kg, category)
````
